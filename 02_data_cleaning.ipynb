```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/content/drive/MyDrive/weather_preddiction/data/open_meteo.csv')
```

```python
data.isna().sum()
```

```python
data['time'] = pd.to_datetime(data['time'])
```

```python
data.head()
```

```python
data['year'] = data['time'].dt.year
data['month'] = data['time'].dt.month
data['day'] = data['time'].dt.day
data['hour'] = data['time'].dt.hour
```

```python
data.head()
```

```python
data.describe().round(1)
```

```python
data.info()
```

```python
data.drop(['cloud_cover_low (%)', 'cloud_cover_mid (%)', 'cloud_cover_high (%)', 'wind_speed_100m (km/h)', 'wind_direction_100m (°)', 'wind_gusts_10m (km/h)'], axis=True)
```

```python
data.drop_duplicates()
```
